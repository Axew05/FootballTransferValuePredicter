```python
import pandas as pd

raw = "../data-raw"

players = pd.read_csv(f"{raw}/players.csv.gz")
vals = pd.read_csv(f"{raw}/player_valuations.csv.gz", parse_dates=["date"])
games = pd.read_csv(f"{raw}/games.csv.gz", parse_dates=["date"])
apps = pd.read_csv(f"{raw}/appearances.csv.gz", parse_dates=["date"])
clubs = pd.read_csv(f"{raw}/clubs.csv.gz")

#print(games["competition_id"].value_counts().head(10))
#pl = games[games["competition_id"] == "GB1"]
#print(pl["season"].value_counts().sort_index())

#print(vals["date"].min(), vals["date"].max())
#print(vals["market_value_in_eur"].describe())
#print(vals.groupby(vals["date"].dt.year).size())

peak = (
    vals.groupby("player_id")["market_value_in_eur"]
    .max()
    .sort_values(ascending=False)
    .head(20)
    .reset_index()
)

top20 = peak.merge(players[["player_id", "first_name", "last_name"]], on="player_id", how="left")
top20["full_name"] = top20["first_name"].fillna("") + " " + top20["last_name"].fillna("")
top20["full_name"] = top20["full_name"].str.strip()

top20[["full_name", "market_value_in_eur"]]

#print(players[["date_of_birth", "position", "contract_expiration_date"]].isna().mean())
```
